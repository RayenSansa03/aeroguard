```python
import h5py

chemin = "/content/raw/N-CMAPSS_DS01-005.h5"

with h5py.File(chemin, "r") as f:
    for tiroir in f.keys():
        print(f"{tiroir:10} → forme {f[tiroir].shape}")
```

```python
import numpy as np

def lire_noms(f, nom):
    """Lit un tiroir de noms (_var) et renvoie une liste de textes lisibles."""
    valeurs = f[nom][()]
    return [v.decode() if isinstance(v, bytes) else str(v) for v in valeurs]

with h5py.File(chemin, "r") as f:
    for nom in ["W_var", "X_s_var", "X_v_var", "T_var", "A_var"]:
        colonnes = lire_noms(f, nom)
        print(f"{nom:8} ({len(colonnes)}) : {colonnes}")
```
